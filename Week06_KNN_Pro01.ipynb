```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsRegressor
```

```python
# 1. Data Setup
study_hours = np.array([1,2,3,3.5,4,5,6,7]).reshape(-1,1)
scores = np.array([20,30,40,80,50,60,70,80])
```

```python
# 2. Model Training
model = KNeighborsRegressor(n_neighbors=6)
model.fit(study_hours, scores)
plt.scatter(study_hours, scores, color='blue', label='Original Data (Before)', s=100)
```

```python
# 3. Visaluixation
plt.figure(figsize=(10,6))
# Before: Plot the training data points
plt.scatter(study_hours, scores, color='blue', label='Original Data (Before)', s=100)
```

```python
# 4. Testing a Specific point (4.5 hours)
test_point = 2.5
test_pred = model.predict([[test_point]])
plt.scatter(test_point, test_pred, color='green', marker='X', s=200, label=f'Test Poit: {test_point}h')
```

```python
# Formatting the Plot
plt.title('KNN Regression: Study Hours vs. Test Scores')
plt.xlabel('Study Hours')
plt.ylabel('Test Scores')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
